```
import numpy as np
import pandas as pd
```

```
p=0.9
mat = pd.read_csv("data.csv", header=None)
mat
```

```
cov_mat=np.cov(mat.T)
cov_mat
```

```
eigen_vals, eigen_vecs = np.linalg.eig(cov_mat)
eigen_vecs = eigen_vecs / np.linalg.norm(eigen_vecs, axis=0, keepdims=True)

sorted_idx = np.argsort(eigen_vals)[::-1]
eigen_vals = eigen_vals[sorted_idx]
eigen_vecs = eigen_vecs[:, sorted_idx]
eigen_vals,eigen_vecs
```

```
s=0
num_pcs=0
for i in range(len(eigen_vals)):
    s+=eigen_vals[i]
    if s/sum(eigen_vals) >= p:
        num_pcs=i+1
        break


num_pcs
```

```
principal_vecs = eigen_vecs[:, :num_pcs]

X=mat.to_numpy()
X=X-X.mean(axis=0)
new_pcs = X @ principal_vecs

new_pcs = pd.DataFrame(
    new_pcs,
    columns=[f"PC{i + 1}" for i in range(num_pcs)]
)
new_pcs
```
